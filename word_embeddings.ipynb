```python
!pip install gensim -q #Word2vec
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from gensim.models import Word2Vec
```

```python
corpus = [
    "machine learning is fun",
    "machine learning is powerful",
    "deep learning drives ai",
    "natural language processing is interesting"
]

corpus #Document collection
```

```python
#Vocabulary of the document
vocab = sorted(list(set(" ".join(corpus).split())))
print("Vocabulary:", vocab)
```

```python
#Bag of words
bow = CountVectorizer()
bow_matrix = bow.fit_transform(corpus)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow.get_feature_names_out()
)

bow_df
```

```python
corpus2 = [
    "students learn python",
    "students learn machine learning",
    "python is easy to learn",
    "machine learning is useful",
    "students use python for projects"
]
```

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(corpus)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

tfidf_df
```

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(corpus2)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

tfidf_df
```
