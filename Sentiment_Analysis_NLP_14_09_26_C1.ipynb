```python
import re
import string
import nltk #Natural Language tool kit

# Download required NLTK resources (run once)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
print("Libraries loaded successfully")
```

```python
text = "I absolutely LOVE this movie!!! It's amazing, but the ending was not good. 😍 #BestMovie2026"
print("Original Text:", text)
```

```python
#1. converting into lowercase

text_lower = text.lower()

print("\n1. Lowercase:")
print(text_lower)
```

```python
text_no_urls = re.sub(r'http\S+|www\S+', '', text_lower)

print("\n2. Remove URLs:")
print(text_no_urls)
```

```python
text_no_mentions = re.sub(r'@\w+', '', text_no_urls)
text_no_hashtags = re.sub(r'#', '', text_no_mentions)

print("\n3. Remove Mentions and Hashtags:")
print(text_no_hashtags)
```

```python
text_no_emojis = text_no_hashtags.encode(
    'ascii', 'ignore'
).decode('ascii')

print("\n4. Remove Emojis:")
print(text_no_emojis)
```

```python
# 6. Remove Punctuation
text_no_punctuation = text_no_emojis.translate(
    str.maketrans('', '', string.punctuation)
)

print("\n5. Remove Punctuation:")
print(text_no_punctuation)

# 7. Tokenization
tokens = word_tokenize(text_no_punctuation)

print("\n6. Tokenization:")
print(tokens)

# 8. Stopword Removal
stop_words = set(stopwords.words('english'))

# Preserve important sentiment negation words
negation_words = {'not', 'no', 'never', 'neither', 'nor'}

filtered_tokens = [
    word for word in tokens
    if word not in stop_words or word in negation_words
]

print("\n7. Stopword Removal:")
print(filtered_tokens)

# 9. Stemming


stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word) for word in filtered_tokens
]

print("\n8. Stemming:")
print(stemmed_words)

# 10. Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word) for word in filtered_tokens
]

print("\n9. Lemmatization:")
print(lemmatized_words)

#11. Vectorization
cleaned_text = ' '.join(lemmatized_words)

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform([cleaned_text])

print("\n10. Cleaned Text for TF-IDF:")
print(cleaned_text)

print("\nTF-IDF Feature Names:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Numerical Vector:")
print(tfidf_matrix.toarray())
```

#Ruled based approach

```python
import pandas as pd
data = {
    "text": [
        "I love this product, it is amazing",
        "The movie was excellent and enjoyable",
        "I hate this service, it is terrible",
        "The product is bad and disappointing",
        "The experience was wonderful and fantastic"
    ],
    "sentiment": [
        "positive",
        "positive",
        "negative",
        "negative",
        "positive"
    ]
}
df=pd.DataFrame(data)
print(df)
```

```python
#Creating a dictionary of positive and negative words

positive_words = {
    "love", "amazing", "excellent", "enjoyable",
    "wonderful", "fantastic", "good", "great",
    "happy", "best"
}

negative_words = {
    "hate", "terrible", "bad", "disappointing",
    "poor", "worst", "sad", "awful"
}
```

```python
def rule_based_sentiment(text):
    # Convert text to lowercase
    text = text.lower()

    # Extract words
    words = re.findall(r'\b\w+\b', text)
    print("Extracted words", words)
    # Count positive and negative words
    positive_count = sum(word in positive_words for word in words)
    negative_count = sum(word in negative_words for word in words)

    if positive_count>negative_count:
      return "postive"
    elif negative_count> positive_count:
      return "negative"
    else:
      return "neutral"
```

```python
df['rule_based_prediction']=df["text"].apply(rule_based_sentiment)

print("Final results")
print(df)
```

```python
X = df["text"]
y = df["sentiment"]

vectorizer = TfidfVectorizer()

X_tfidf = vectorizer.fit_transform(X)

print("\nTF-IDF Feature Matrix Shape:")
print(X_tfidf.shape)
```

```python
tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=vectorizer.get_feature_names_out()
)
tfidf_df
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
#Model Training
model.fit(X_tfidf, y)
# 8. Predict Sentiment for Dummy Dataset

ml_predictions = model.predict(X_tfidf)

df["ml_prediction"] = ml_predictions

print("\nMachine Learning Results:")
print(df[["text", "sentiment", "ml_prediction"]])
```
